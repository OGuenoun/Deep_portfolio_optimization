# Projet 4 — Comparison of all models

Loads every run in `models/*/results/*/` and evaluates them on the same basis.

## 1. Load the dataset
`Y` (oracle weights), `R_fwd` (forward returns), `meta` (split, label type) from `data/processed/N63_wmax1`.

## 2. Load all experiments
One entry per `models/<model>/results/<run_name>/`: `weights_val.csv`, `weights_test.csv`, `config.json`.

## 3. Fit to the oracle labels
Error between predicted weights and `Y` (MSE, MAE), share of correct cash / invested calls.

## 4. Backtest
Rebalancing every N days. Excess return, volatility, Sharpe, max drawdown, turnover.
Benchmarks: oracle (upper bound), equal weight, 100% cash, rolling Markowitz.

## 5. Ablations
Horizon N = 21 vs 63, weight cap 25% vs none.
